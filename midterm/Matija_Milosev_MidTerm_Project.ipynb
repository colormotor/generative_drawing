## Technical Implementation

I started with iterations of the python sketches we worked on during the labs and then implemented those in addition to my own experiments. 

For the physical production of the work I used a pen plotting machine to plot the lines which were generated by code and saved as svg files. After this I had a process of experimenting with different mediums, this included different pens, a variety of papers and surfaces. Aditionally I used different parameters like speed and positioning of the plotter in the axidraw plugin within the inkscape software. 

Ultimatelly I decided to use a white marker and white pen on black background because of the contrast this achieved. 
Especially knowing the work was being showed in a space with white walls this decision seemed the most interesting and enticing to myself. 

Due to the nature of overlaying sketches within software and then plotting them over each other I also decided to play with this element and overlay some of the papers onto the final work creating a collage.
